```python
import pandas as pd
import matplotlib.pyplot as plt
from datetime import datetime, timedelta
```

```python
metadata_df = pd.read_csv("data/metadata.csv")
processdata_df = pd.read_csv("data/processdata.csv")
productiondata_df = pd.read_csv("data/productiondata.csv")
```

```python
print(metadata_df.head())
```

```python
print(metadata_df['Recipe'].unique())

metadata_df['Recipe'].value_counts().plot(kind='bar')

plt.xlabel('Recipe')
plt.ylabel('Count')
plt.title('Recipe Distribution')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
print(metadata_df['Description'].unique())

metadata_df['Description'].value_counts().plot(kind='bar')

plt.xlabel('Description')
plt.ylabel('Count')
plt.title('Description Distribution')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
